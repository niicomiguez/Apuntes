```sage
##EJERCICIO 2
def aproximarError(error):
    contador=0
    n=1
    valorFuncion=0
    ejecutar=True
    while(ejecutar):
        valorFuncion=1/(n+1)
        contador+=1
        if(valorFuncion<=error):
            ejecutar=False
        n+=1
    return contador

show(aproximarError(0.05))
```

```sage
##EJERCICIO 4
def multiplicacionRusa(a, b):
    resultadosA = []
    resultadosB = []
    sumandos = []
    b_actual = b
    while a >= 1:
        resultadosA.append(a)
        resultadosB.append(b_actual)
        if a % 2 != 0:
            sumandos.append(b_actual)
        else:
            sumandos.append(0)
        a = a // 2
        b_actual = b_actual * 2
    show(resultadosA)
    show(resultadosB)
    show(sumandos)
    return sum(sumandos)

print(multiplicacionRusa(27, 82))
    
```

```sage
##EJERCICIO 5
def numerosPrimos(valor):
    lista_primos=[]
    for p in primes(valor):
        if (p-1)%3==0:
            lista_primos.append(p)
    return lista_primos
show(numerosPrimos(50))
```
